```python
#1.How do you create a 2D NumPy array and calculate the sum of each row?
import numpy as np
a=np.arange(5,50,3)
print(a)
a=a.reshape(3,5)
print(a)

#2.Write a Pandas script to find the mean of a specific column in a DataFrome.
import pandas as pd
data=({'name':( "raju","geetha","ravali","ramesh"),
    'marks': (95,90,25,69)})
df = pd.DataFrame(data)
print(df)

#3.Create a scatter plot using Matplotlib.
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
plt.scatter(df['name'],df['marks'])
plt.title("Student-Marks Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.show()

#4.How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap.
print(a) # ‘a’ from first program
corr_matrix = np.corrcoef(a)
# (Optional) Convert to DataFrame for easier handling
df_corr_matrix = pd.DataFrame(corr_matrix)
plt.figure(figsize=(10, 8))
sns.heatmap(df_corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt.show()

#5.Generate a bar plot using Plotly.
import plotly.express as px
import warnings
warnings.filterwarnings("ignore")
fig = px.bar(df, x='name', y='marks', color='name', title='Student-Marks Chart', labels={'name': 'Students', 'marks': 'Marks'}, color_continuous_scale='Viridis' )
fig.show()

#6.Create a DataFrame and add a new column based on an existing column.
print(df) # from 2nd program
def assign_grade(marks):

    if (marks>75 and marks<=100):
        return 'A'
    elif (marks>60 and marks<75):
        return 'B'
    elif (marks>35 and marks<60):
        return 'C'

    else:
        return 'Fail'
df['Grade'] = df['marks'].apply(assign_grade) # Display the DataFrame
print(df)

#7.Write a program to perform element—wise multiplication of two NumPy arrays.
array1 = np.array([1, 2, 3, 4])
array2 = np.array([5, 6, 7, 8])
# Perform element-wise multiplication
result = array1 * array2
print("Result of element-wise multiplication:", result)

#8.Create a line plot with multiple lines using Matplotlib.
import warnings
warnings.filterwarnings("ignore")
array3=np.array([5,7,2,8])
plt.plot(array3, array1, label='Line 1', marker='o')
plt.plot(array3, array2, label='Line 2', marker='s')
plt.title("Student-Marks Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.show()

#9.Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold.
threshold = 70
# Filter rows where 'marks' is greater than the threshold
df2 = df[df['marks'] > threshold]
print(df2)

#10.Create a histogram using Seaborn to visualize a distribution.
dataa = np.random.randn(1000)
plt.figure(figsize=(10, 6))
sns.histplot(dataa, bins=30, kde=True, color='blue')
plt.title('Histogram of Normally Distributed Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

#11.Perform matrix multiplication using NumPy.
array1=array1.reshape(2,2)
array2=array2.reshape(2,2)
print(array1,array2)
print(array1@array2)

#12.Use Pandas to load a CSV file and display its first 5 rows.
#df1=pd.read_csv('Auto.csv')
#print(df1.head())

#13.Create a 3D scatter plot using Plotly.
fig = px.scatter_3d(df, x='name', y='marks', z='Grade', color='name', title='Student-Marks 3D Chart', labels={'name': 'Students', 'marks': 'Marks', 'Grade': 'Grade'})
fig.show()
```
